# Check whether a drug needs FDA label curation

This deterministic preflight checks whether an FDA application already has a MOAlmanac document and, if so, whether openFDA lists a newer approved label.

```python
from pathlib import Path

from moalmanac_fda_curation.core.check_curation_preflight import check_curation_preflight
```

## Configure the check

Use the NDA, BLA, or ANDA application number from Drugs@FDA. The database path points to a neighboring `moalmanac-db` checkout.

```python
APPLICATION_NUMBER = "BLA125554"
PROJECT_ROOT = next(
    path
    for path in (Path.cwd(), *Path.cwd().parents)
    if (path / "pyproject.toml").exists()
)
DOCUMENTS_JSON = PROJECT_ROOT.parent / "moalmanac-db/referenced/documents.json"
```

## Run the preflight

If `previously_curated` is false, this application belongs in the new-drug workflow. If both booleans are true, continue to the revised- and new-indication checks.

```python
preflight = check_curation_preflight(APPLICATION_NUMBER, DOCUMENTS_JSON)
preflight
```
